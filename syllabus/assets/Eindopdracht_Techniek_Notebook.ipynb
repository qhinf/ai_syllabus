```python
# Importeer alle benodigde bibliotheken
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Voor het AI-model
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Zorgt ervoor dat grafieken mooi in het notebook verschijnen
%matplotlib inline

# Maak grafieken mooier
sns.set_style('whitegrid')

print("Alle bibliotheken zijn geladen!")
```

Hier laden we de dataset in. Pas de bestandsnaam aan naar jouw eigen dataset!

**LET OP:** Zorg dat je dataset-bestand in dezelfde map staat als dit notebook, of geef het volledige pad op.

```python
# Laad je dataset in (pas de bestandsnaam aan!)
# Voorbeeld: als je bestand 'spotify_data.csv' heet:
df = pd.read_csv('jouw_dataset.csv')

# Laat de eerste 5 rijen zien
print("De eerste 5 rijen van de data:")
df.head()
```

## Stap 3: Data verkennen

Voordat we met AI beginnen, moeten we eerst begrijpen wat voor data we hebben.

```python
# Bekijk de grootte van de dataset
print(f"Aantal rijen: {df.shape[0]}")
print(f"Aantal kolommen: {df.shape[1]}")
print("\n" + "="*50 + "\n")

# Bekijk informatie over de kolommen
print("Informatie over de kolommen:")
df.info()
```

## Stap 4: Data visualiseren

Grafieken helpen ons om patronen in de data te zien.

```python
# Maak een grafiek van de verdeling van je belangrijkste variabele
# PAS DIT AAN naar jouw eigen kolommen!

plt.figure(figsize=(10, 6))

# Voorbeeld: verdeling van populariteit
# Vervang 'popularity' door jouw kolomnaam
plt.hist(df['kolom_naam'], bins=30, edgecolor='black')
plt.xlabel('Jouw variabele')
plt.ylabel('Aantal')
plt.title('Verdeling van [jouw variabele]')
plt.show()
```

## Stap 5: Data voorbereiden voor AI

AI-modellen hebben specifieke input nodig. We moeten onze data voorbereiden.

```python
# BELANGRIJK: Pas deze variabelen aan voor jouw onderzoeksvraag!

# Features (X) = de eigenschappen die we gebruiken om te voorspellen
# Bijvoorbeeld: tempo, energie, dansability
feature_columns = ['kolom1', 'kolom2', 'kolom3']  # PAS DIT AAN!
X = df[feature_columns]

# Target (y) = wat we willen voorspellen
# Bijvoorbeeld: is_hit (1 = hit, 0 = geen hit)
target_column = 'jouw_target_kolom'  # PAS DIT AAN!
y = df[target_column]

print(f"Features shape: {X.shape}")
print(f"Target shape: {y.shape}")
```

```python
# Splits de data in trainingsdata en testdata
# 80% voor training, 20% voor testen

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,  # 20% voor testen
    random_state=42  # Voor reproduceerbare resultaten
)

print(f"Training set: {X_train.shape[0]} voorbeelden")
print(f"Test set: {X_test.shape[0]} voorbeelden")
print("\nData is succesvol gesplitst!")
```

## Stap 6: AI-Model trainen

Nu gaan we het AI-model trainen! We gebruiken een **Random Forest**, een type AI dat goed werkt voor beginners.

**Hoe werkt Random Forest?**
- Het maakt meerdere 'beslissingsbomen' die elk een voorspelling doen
- De uiteindelijke voorspelling is de meerderheid van alle bomen
- Het kan goed omgaan met complexe data

```python
# Maak het AI-model aan
model = RandomForestClassifier(
    n_estimators=100,  # Aantal beslissingsbomen
    random_state=42,
    max_depth=10  # Maximale diepte van elke boom
)

# Train het model met de trainingsdata
print("Model wordt getraind...")
model.fit(X_train, y_train)
print("Model is succesvol getraind!")
```

## Stap 7: Model evalueren

Nu gaan we kijken hoe goed ons model werkt!

```python
# Laat het model voorspellingen doen op de testdata
y_pred = model.predict(X_test)

# Bereken de nauwkeurigheid (accuracy)
accuracy = accuracy_score(y_test, y_pred)

print(f"\n{'='*50}")
print(f"Model Nauwkeurigheid: {accuracy:.2%}")
print(f"{'='*50}\n")

# Wat betekent dit?
print(f"Het model voorspelt {accuracy:.1%} van de gevallen correct!")
```

```python
# Gedetailleerd rapport
print("Gedetailleerd classificatierapport:")
print(classification_report(y_test, y_pred))

# Uitleg van de termen:
# - Precision: van alle voorspelde hits, hoeveel zijn echt hits?
# - Recall: van alle echte hits, hoeveel heeft het model gevonden?
# - F1-score: combinatie van precision en recall
```

```python
# Confusion Matrix - laat zien waar het model fouten maakt
cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.xlabel('Voorspelde waarde')
plt.ylabel('Echte waarde')
plt.title('Confusion Matrix')
plt.show()

# Uitleg:
# - Linksboven: correct voorspeld als negatief
# - Rechtsboven: foutief voorspeld als positief
# - Linksonder: foutief voorspeld als negatief
# - Rechtsonder: correct voorspeld als positief
```

## Stap 8: Feature Importance

Welke eigenschappen zijn het belangrijkst voor de voorspelling?

```python
# Haal de belangrijkheid van elke feature op
feature_importance = pd.DataFrame({
    'feature': feature_columns,
    'importance': model.feature_importances_
}).sort_values('importance', ascending=False)

# Maak een grafiek
plt.figure(figsize=(10, 6))
plt.barh(feature_importance['feature'], feature_importance['importance'])
plt.xlabel('Belangrijkheid')
plt.title('Welke eigenschappen zijn het belangrijkst?')
plt.gca().invert_yaxis()
plt.show()

print("\nTop 3 belangrijkste eigenschappen:")
print(feature_importance.head(3))
```
